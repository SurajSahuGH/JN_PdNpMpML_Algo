```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# Resize your Graph
plt.figure(figsize=(5,3), dpi=100)

# Line 1

# Keyword Argument Notation
# plt.plot(x,y, label='2x', color='red', linewidth=2, linestyle='--', marker='.', markersize=10, markeredgecolor='black')

# shorthand notation
# fmt = '[color][marker][line]'
plt.plot(x,y, 'b.--', label='2x')

# Line 2

# select interval we want to plot points at
x2 = np.arange(0,4.5,0.5)

# plot part of graph as line
plt.plot(x2[:6], x2[:6]**2,'r', label='X^2')

# plot remainder of graph as a dot
plt.plot(x2[5:], x2[5:]**2,'r--')

# add a title (specify font params with 'fontdict')
plt.title('First Graph', fontdict={'fontname': 'serif','fontsize': 20})

# X and Y labels
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

# X, Y axis tickmarks (scale of your graph)
plt.xticks([0,1,2,3,4,5])
# plt.yticks([0,2,4,6,8,10])

# add a legend
plt.legend()

# save figure (dpi 300 is good when saving so graph has high resolution)
plt.savefig('mygraph.png', dpi=300)

# show plot
plt.show()
```

### Bar Charts

```python
labels = ['A', 'B', 'C']
values = [1,4,2]

bars = plt.bar(labels, values)

# bars[0].set_hatch('/')
# bars[1].set_hatch('O')
# bars[2].set_hatch('X')

patterns = ['/','O','X']
for bar in bars:
    bar.set_hatch(patterns.pop(0))

plt.figure(figsize=(6,4))

plt.show()
```
