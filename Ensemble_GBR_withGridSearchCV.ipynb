```python
# Import Liberaries,  including GridSearchCV

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn import ensemble
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import GridSearchCV
```

```python
# loading data from csv file

df = pd.read_csv('Melbourne_housing_FULL.csv')
df
```

```python
# Data Scrubbing

# Deleting comparatively unrelevent colunms for predicting house prices from df

del df['Address']
del df['Method']
del df['SellerG']
del df['Date']
del df['Postcode']
del df['Lattitude']
del df['Longtitude']
del df['Regionname']
del df['Propertycount']
```

```python
df
```

```python
# removing rows with any empty cell (missing values)

df.dropna(axis=0, how='any', subset=None, inplace=True)
```

```python
df
```

```python
# Using one hot encoding for convt non numeric to numeric

df = pd.get_dummies(df, columns = ['Suburb', 'CouncilArea', 'Type'])
```

```python
df
```

```python
# assinging X as independent vars and y as dependent var

X = df.drop('Price', axis=1)
y = df['Price']
```

```python
# splitting dataset into training and test segms (70/30) & shuffling dataset

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, shuffle = True)
```

```python
# input algo

model = ensemble.GradientBoostingRegressor()

#    n_estimators = 250,    # no. of decision trees
#    learning_rate = 0.1,   # effectively shrinks the contribution of each tree 
#   max_depth = 5,        # max no. of layers (depth) for each tree
#    min_samples_split = 4, # min no. of samples req to exec a new binary split
#    min_samples_leaf = 6,  # -|- req within each leaf for a new branch to be created
#    max_features = 0.6,    # total no. of features presented for determn the best split
#   loss = 'huber'         # model's error rate
```

```python
# Set the configurations that you wish to test. To minimize processing time, limit num. of variables or
# experiment on each hyperparameter separately.

hyperparameters = {
'n_estimators': [200, 300],
'max_depth': [4, 6],
'min_samples_split': [3, 4],
'min_samples_leaf': [5, 6],
'learning_rate': [0.01, 0.02],
'max_features': [0.8, 0.9],
'loss': ['ls', 'lad', 'huber']
}

```

```python
# Define grid search. Run with four CPUs in parallel if applicable.

grid = GridSearchCV(model, hyperparameters, n_jobs = 4)
```

```python
# linking the training data to the learning algo (feeding training data to algo)
# run grid search on training data

grid.fit(X_train, y_train)
```

```python
# Return optimal hyperparameters
grid.best_params_
```

```python
# evaluating the result by MAE up to two decimal places

mae_train = mean_absolute_error(y_train, grid.predict(X_train))
# print(f"Training Set Mean Absolute Error:{mae_train:.2f}")
print("Training Set Mean Absolute Error:%.2f" % mae_train)
```

```python
# 2 

mae_test = mean_absolute_error(y_test, grid.predict(X_test))
print("Test Set Mean Absolute Error:%.2f" % mae_test)
```
